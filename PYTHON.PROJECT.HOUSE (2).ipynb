# OUTLIER

```python
# Calculating Q1, Q3 & IQR for sqft

Q1 = df['sqft_living'].quantile(0.25)
Q3 = df['sqft_living'].quantile(0.75)
IQR = Q3 - Q1


Lower_bound = Q1 - 1.5 * IQR
Upper_bound = Q3 + 1.5 * IQR


sqft_outliers = df[(df['sqft_living'] < Lower_bound) | (df['sqft_living'] > Upper_bound)]


print("=== SQFT OUTLIERS USING IQR ===")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", Lower_bound)
print("Upper Bound:", Upper_bound)
print("Total Outliers:", sqft_outliers.shape[0])
```

```python
# Calculate Q1, Q3 & IQR for bedrooms

Q1 = df['bedrooms'].quantile(0.25)
Q3 = df['bedrooms'].quantile(0.75)
IQR = Q3 - Q1


Lower_bound = Q1 - 1.5 * IQR
Upper_bound = Q3 + 1.5 * IQR



bedroom_outliers = df[(df['bedrooms'] < Lower_bound) | (df['bedrooms'] > Upper_bound)]

# Output summary
print("=== BEDROOMS OUTLIERS USING IQR ===")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", Lower_bound)
print("Upper Bound:", Upper_bound)
print("Total Outliers:", bedroom_outliers.shape[0])
```

```python
import pandas as pd
from scipy import stats


sample_data = df["price"]

population_mean = 600000

t_statistic, p_value = stats.ttest_1samp(sample_data, population_mean)

print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value:}")    # show more precision

alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

```python
from scipy.stats import chi2_contingency


contingency_table=pd.crosstab(df['bedrooms'], df['condition'])
#Chi-Square Test

chi2, p_values, dof, expected =chi2_contingency(contingency_table)

print("Contingency Table:\n", contingency_table) 
print("\n chi-Square Statstic", chi2)

print("p-value:", p_values)
print("Degrees of Freedom:", dof)

print("\nExpected Frequencies: \n", pd.DataFrame (expected, contingency_table.index,contingency_table.columns))

if p_values < 0.05: 
    print("\n Reject if There is a significant relationship between bedrooms and condition")
else:
    print("\n Fail to Reject the Null Hypothesis: There is NO significant relationship between the variables.")
```
